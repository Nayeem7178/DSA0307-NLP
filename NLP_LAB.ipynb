EXP 02- Implement a basic finite state automaton that recognizes a specific language or pattern. In this
example, we'll create a simple automaton to match strings ending with 'ab' using python


```python
def fsa(string):
    state = 0

    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

s = input("Enter a string: ")
fsa(s)
```

11. Implement a simple top-down parser for context-free grammars using python.

### Top-Down Parser for Context-Free Grammars

This program implements a basic top-down parser. It attempts to derive a given input sentence from a defined context-free grammar by recursively expanding non-terminal symbols until terminals match the input.

```python
class Parser:
    def __init__(self, grammar):
        self.grammar = grammar
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0
        return self.parse_symbol(self.grammar['S'])

    def parse_symbol(self, productions):
        original_index = self.current_token_index
        for production in productions:
            self.current_token_index = original_index
            if self.match_production(production):
                return True
        return False

    def match_production(self, production):
        for symbol in production:
            if symbol in self.grammar:  # Non-terminal
                if not self.parse_symbol(self.grammar[symbol]):
                    return False
            else:  # Terminal
                if not self.match_terminal(symbol):
                    return False
        return True

    def match_terminal(self, terminal):
        if self.current_token_index < len(self.tokens) and self.tokens[self.current_token_index] == terminal:
            self.current_token_index += 1
            return True
        return False

# Define a simple context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test cases
sentences = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog",
    "chased the dog"
]

parser = Parser(grammar)

print("Parsing Results:")
for sentence in sentences:
    if parser.parse(sentence) and parser.current_token_index == len(parser.tokens):
        print(f"'{sentence}' -> Accepted")
    else:
        print(f"'{sentence}' -> Rejected")
```

12. Implement an Earley parser for context-free grammars using a simple python program.

### Earley Parser for Context-Free Grammars

An Earley parser is a chart parser that uses dynamic programming to parse sentences according to a context-free grammar. It's known for its ability to handle all context-free grammars, including ambiguous and left-recursive ones.

```python
class EarleyParser:
    def __init__(self, grammar, start_symbol='S'):
        self.grammar = grammar
        self.start_symbol = start_symbol
        self.chart = []

    def parse(self, words):
        n = len(words)
        self.chart = [[] for _ in range(n + 1)]

        # Initialize chart[0] with the start symbol
        for prod in self.grammar.get(self.start_symbol, []):
            self.chart[0].append((self.start_symbol, prod, 0, 0))

        for i in range(n + 1):
            for item in self.chart[i]:
                rule_lhs, rule_rhs, dot_pos, origin = item

                if dot_pos < len(rule_rhs):  # Not a complete item
                    next_symbol = rule_rhs[dot_pos]
                    if next_symbol in self.grammar:  # Non-terminal
                        self.predict(next_symbol, i)
                    else:  # Terminal
                        self.scan(item, words, i)
                else:  # Complete item
                    self.complete(item, i)

        # Check if the sentence is accepted
        for item in self.chart[n]:
            if item[0] == self.start_symbol and item[2] == len(item[1]) and item[3] == 0:
                return True
        return False

    def predict(self, non_terminal, j):
        for prod in self.grammar.get(non_terminal, []):
            new_item = (non_terminal, prod, 0, j)
            if new_item not in self.chart[j]:
                self.chart[j].append(new_item)

    def scan(self, item, words, j):
        rule_lhs, rule_rhs, dot_pos, origin = item
        next_symbol = rule_rhs[dot_pos]
        if j < len(words) and words[j] == next_symbol:
            new_item = (rule_lhs, rule_rhs, dot_pos + 1, origin)
            if new_item not in self.chart[j + 1]:
                self.chart[j + 1].append(new_item)

    def complete(self, item, j):
        rule_lhs, rule_rhs, dot_pos, origin = item
        for prev_item in self.chart[origin]:
            prev_lhs, prev_rhs, prev_dot_pos, prev_origin = prev_item
            if prev_dot_pos < len(prev_rhs) and prev_rhs[prev_dot_pos] == rule_lhs:
                new_item = (prev_lhs, prev_rhs, prev_dot_pos + 1, prev_origin)
                if new_item not in self.chart[j]:
                    self.chart[j].append(new_item)

# Define a simple context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test cases
sentences = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog",
    "chased the dog",
    "a cat"
]

earley_parser = EarleyParser(grammar)

print("Earley Parser Results:")
for sentence in sentences:
    words = sentence.split()
    if earley_parser.parse(words):
        print(f"'{sentence}' -> Accepted")
    else:
        print(f"'{sentence}' -> Rejected")
```

13. Generate a parse tree for a given sentence using a context-free grammar using python program.

### Generating a Parse Tree from a Context-Free Grammar

This program implements a recursive descent parser with backtracking to construct a parse tree for a given sentence based on a context-free grammar. The `ParseTreeNode` class helps in representing the tree structure.

```python
class ParseTreeNode:
    def __init__(self, label, children=None):
        self.label = label
        self.children = children if children is not None else []

    def __str__(self, level=0):
        ret = "  " * level + self.label + "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.children})"


class RecursiveDescentParserWithTree:
    def __init__(self, grammar, start_symbol='S'):
        self.grammar = grammar
        self.start_symbol = start_symbol
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0

        root_node = self._parse_non_terminal(self.start_symbol)

        if root_node and self.current_token_index == len(self.tokens):
            return root_node
        else:
            return None

    def _parse_non_terminal(self, non_terminal_symbol):
        original_token_index = self.current_token_index

        for production in self.grammar.get(non_terminal_symbol, []):
            self.current_token_index = original_token_index # Backtrack
            children_nodes = []
            match_success = True

            for symbol in production:
                if symbol in self.grammar:  # Is a non-terminal
                    child_node = self._parse_non_terminal(symbol)
                    if child_node:
                        children_nodes.append(child_node)
                    else:
                        match_success = False
                        break
                else:  # Is a terminal
                    child_node = self._parse_terminal(symbol)
                    if child_node:
                        children_nodes.append(child_node)
                    else:
                        match_success = False
                        break

            if match_success:
                return ParseTreeNode(non_terminal_symbol, children_nodes)

        return None # No production matched

    def _parse_terminal(self, terminal_symbol):
        if self.current_token_index < len(self.tokens) and \
           self.tokens[self.current_token_index] == terminal_symbol:
            node = ParseTreeNode(terminal_symbol)
            self.current_token_index += 1
            return node
        return None

# Define the context-free grammar (same as before)
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test sentences
sentences_for_tree = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog" # This sentence will likely not form a complete 'S' tree
]

r_parser = RecursiveDescentParserWithTree(grammar)

print("Parse Tree Generation Results:")
for sentence in sentences_for_tree:
    print(f"\nSentence: '{sentence}'")
    parse_tree = r_parser.parse(sentence)
    if parse_tree:
        print("Accepted. Parse Tree:")
        print(parse_tree)
    else:
        print("Rejected. No complete parse tree found.")
```

14. Create a program in python to check for agreement in sentences based on a context-free
grammar's rules.

### Agreement Checking in Context-Free Grammars

This program implements a recursive descent parser capable of checking for feature agreement (e.g., subject-verb number agreement). The grammar and lexicon are extended to include features for symbols, and the parser enforces these constraints during parsing.

```python
class ParseTreeNode:
    def __init__(self, label, children=None, features=None):
        self.label = label
        self.children = children if children is not None else []
        self.features = features if features is not None else {}

    def __str__(self, level=0):
        ret = "  " * level + self.label
        if self.features:
            ret += f" {self.features}"
        ret += "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.features}, {self.children})"


class FeatureParser:
    def __init__(self, grammar, lexicon, start_symbol='S'):
        self.grammar = grammar
        self.lexicon = lexicon
        self.start_symbol = start_symbol
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0

        # Try to parse the start symbol
        result = self._parse_non_terminal(self.start_symbol)

        # A successful parse means we got a tree and consumed all tokens
        if result and self.current_token_index == len(self.tokens):
            return result[0]  # Return the ParseTreeNode
        else:
            return None

    def _parse_non_terminal(self, non_terminal_symbol):
        original_token_index = self.current_token_index

        for production_tuple in self.grammar.get(non_terminal_symbol, []):
            production_rule, agreement_func = production_tuple
            self.current_token_index = original_token_index  # Backtrack

            matched_children = []
            child_features = [] # To pass to agreement function
            match_success = True

            for symbol in production_rule:
                if symbol in self.grammar:  # It's a non-terminal
                    child_result = self._parse_non_terminal(symbol)
                    if child_result:
                        child_node, child_feat = child_result
                        matched_children.append(child_node)
                        child_features.append(child_feat)
                    else:
                        match_success = False
                        break
                else:  # It's a terminal (look up in lexicon)
                    child_result = self._parse_terminal(symbol)
                    if child_result:
                        child_node, child_feat = child_result
                        matched_children.append(child_node)
                        child_features.append(child_feat)
                    else:
                        match_success = False
                        break

            if match_success:
                # Apply agreement function if provided
                if agreement_func:
                    # Pass features of children to the agreement function
                    # The agreement function should return parent features or None if agreement fails
                    parent_features = agreement_func(*child_features)
                    if parent_features is None:
                        match_success = False # Agreement failed
                else:
                    parent_features = {}

            if match_success:
                return ParseTreeNode(non_terminal_symbol, matched_children, parent_features), parent_features

        return None

    def _parse_terminal(self, expected_pos_tag):
        # Check if current token matches any word with the expected POS tag
        if self.current_token_index < len(self.tokens):
            current_word = self.tokens[self.current_token_index]
            word_info = self.lexicon.get(current_word)

            if word_info and word_info['POS'] == expected_pos_tag:
                node = ParseTreeNode(current_word, features=word_info)
                self.current_token_index += 1
                return node, word_info
        return None

# --- Define Lexicon and Grammar with Agreement Rules ---

lexicon = {
    'the': {'POS': 'Det', 'NUM': 'any'},
    'a': {'POS': 'Det', 'NUM': 'singular'},
    'cat': {'POS': 'N', 'NUM': 'singular'},
    'cats': {'POS': 'N', 'NUM': 'plural'},
    'dog': {'POS': 'N', 'NUM': 'singular'},
    'dogs': {'POS': 'N', 'NUM': 'plural'},
    'man': {'POS': 'N', 'NUM': 'singular'},
    'men': {'POS': 'N', 'NUM': 'plural'},
    'runs': {'POS': 'V', 'NUM': 'singular'}, # V for singular subject
    'run': {'POS': 'V', 'NUM': 'plural'},   # V for plural subject
    'chases': {'POS': 'V', 'NUM': 'singular'},
    'chase': {'POS': 'V', 'NUM': 'plural'},
    'eats': {'POS': 'V', 'NUM': 'singular'},
    'eat': {'POS': 'V', 'NUM': 'plural'}
}

# Agreement functions:
def s_agreement_check(np_features, vp_features):
    # S inherits number from NP/VP if they agree
    if np_features['NUM'] == vp_features['NUM']:
        return {'NUM': np_features['NUM']}
    return None # Agreement failure

def np_det_n_agreement_check(det_features, n_features):
    # Det must agree with N, or be 'any' (like 'the')
    if det_features['NUM'] == 'any' or det_features['NUM'] == n_features['NUM']:
        return {'NUM': n_features['NUM']}
    return None

def np_n_agreement_check(n_features):
    # NP inherits number from N
    return {'NUM': n_features['NUM']}

def vp_v_np_agreement_check(v_features, np_features):
    # VP inherits number from V (V-O agreement not checked here, only S-V is primary)
    return {'NUM': v_features['NUM']}

def vp_v_agreement_check(v_features):
    # VP inherits number from V
    return {'NUM': v_features['NUM']}

grammar = {
    'S': [
        (['NP', 'VP'], s_agreement_check)
    ],
    'NP': [
        (['Det', 'N'], np_det_n_agreement_check),
        (['N'], np_n_agreement_check)
    ],
    'VP': [
        (['V', 'NP'], vp_v_np_agreement_check), # V with object
        (['V'], vp_v_agreement_check)          # V without object
    ]
}

# Test sentences
sentences_for_agreement = [
    "the cat runs",         # Accepted: 'the' (any), 'cat' (singular), 'runs' (singular)
    "a cat runs",           # Accepted: 'a' (singular), 'cat' (singular), 'runs' (singular)
    "the cats run",         # Accepted: 'the' (any), 'cats' (plural), 'run' (plural)
    "a cats run",           # Rejected: 'a' (singular) with 'cats' (plural)
    "the cat run",          # Rejected: 'cat' (singular) with 'run' (plural)
    "the cats runs",        # Rejected: 'cats' (plural) with 'runs' (singular)
    "men eat",              # Accepted: 'men' (plural), 'eat' (plural)
    "man eats",             # Accepted: 'man' (singular), 'eats' (singular)
    "the dogs chase the cat", # Accepted
    "the dog chases the cats"  # Accepted
]

feature_parser = FeatureParser(grammar, lexicon)

print("Agreement Checking Results:")
for sentence in sentences_for_agreement:
    print(f"\nSentence: '{sentence}'")
    parse_tree = feature_parser.parse(sentence)
    if parse_tree:
        print("Accepted. Parse Tree with Features:")
        print(parse_tree)
    else:
        print("Rejected. Agreement or parsing failed.")
```

15. Implement probabilistic context-free grammar parsing for a sentence using python.

### Probabilistic Context-Free Grammar (PCFG) Parsing

A Probabilistic Context-Free Grammar (PCFG) is a context-free grammar where each production rule has a probability. These probabilities allow the parser to rank different possible parse trees for an ambiguous sentence and choose the most likely one. The CKY (Cocke-Kasami-Younger) algorithm can be adapted to find the most probable parse for a sentence given a PCFG.

```python
import collections

class PCFGParser:
    def __init__(self, pcfg_rules, start_symbol='S'):
        # pcfg_rules is a dictionary: non_terminal -> list of (probability, production_tuple)
        # Example: {'S': [(1.0, ('NP', 'VP'))], ...}
        self.pcfg_rules = pcfg_rules
        self.start_symbol = start_symbol

    def parse(self, sentence_str):
        words = sentence_str.split()
        n = len(words)

        # Initialize the CKY chart
        # chart[i][j] stores a dict: NonTerminal -> (max_prob, backpointer)
        # backpointer = ((split_point, left_NT, right_NT) or (unary_NT,) or word)
        chart = [[collections.defaultdict(lambda: (0.0, None)) for _ in range(n + 1)] for _ in range(n + 1)]

        # Fill in the diagonal (i.e., rules matching single words)
        for i in range(n):
            word = words[i]
            for lhs, productions in self.pcfg_rules.items():
                for prob, rhs in productions:
                    if len(rhs) == 1 and rhs[0] == word: # Terminal production A -> a
                        if prob > chart[i][i+1][lhs][0]:
                            chart[i][i+1][lhs] = (prob, word) # Backpointer is the word itself

        # Fill in the rest of the chart
        for length in range(2, n + 1):  # length of span
            for i in range(n - length + 1):  # start of span
                j = i + length  # end of span

                # Step 1: Process binary rules (A -> B C)
                for k in range(i + 1, j):  # split point
                    for lhs_A, prods_A in self.pcfg_rules.items():
                        for prob_A, rhs_A in prods_A:
                            if len(rhs_A) == 2:  # Binary rule A -> B C
                                B, C = rhs_A
                                prob_B = chart[i][k][B][0] # Probability of B spanning (i, k)
                                prob_C = chart[k][j][C][0] # Probability of C spanning (k, j)

                                if prob_B > 0 and prob_C > 0:
                                    current_prob = prob_A * prob_B * prob_C
                                    if current_prob > chart[i][j][lhs_A][0]:
                                        chart[i][j][lhs_A] = (current_prob, (k, B, C)) # Backpointer for binary rule

                # Step 2: Process unary rules (A -> B) - iterate until no new entries can be added
                changed = True
                while changed:
                    changed = False
                    # Iterate over existing non-terminals in chart[i][j]
                    # We need to iterate over a copy of the keys to allow modification during iteration
                    current_nts_in_cell = list(chart[i][j].keys())

                    for existing_nt in current_nts_in_cell:
                        existing_prob, _ = chart[i][j][existing_nt] # Get prob of existing_nt

                        # Look for unary rules A -> existing_nt
                        for lhs_A, prods_A in self.pcfg_rules.items():
                            for prob_A, rhs_A in prods_A:
                                if len(rhs_A) == 1 and rhs_A[0] == existing_nt: # Unary rule A -> B
                                    # Ensure we don't apply A -> A recursively if it exists
                                    if lhs_A == existing_nt:
                                        continue

                                    current_prob = prob_A * existing_prob
                                    if current_prob > chart[i][j][lhs_A][0]:
                                        chart[i][j][lhs_A] = (current_prob, (existing_nt,)) # Backpointer for unary rule (tuple of 1)
                                        changed = True

        # Reconstruct the most probable parse tree if a parse exists for the start symbol
        max_prob, backpointer = chart[0][n][self.start_symbol]
        if max_prob > 0:
            return max_prob, self._reconstruct_tree(self.start_symbol, 0, n, chart)
        return 0.0, None

    def _reconstruct_tree(self, lhs, i, j, chart):
        max_prob, backpointer = chart[i][j][lhs]

        if isinstance(backpointer, str):  # Terminal node (backpointer is the word)
            # The prob attribute for the ParseTreeNode of the non-terminal (lhs) should reflect
            # the probability of the rule application itself.
            # The actual word node will have a prob=1.0 as it's a leaf.
            return ParseTreeNode(lhs, [ParseTreeNode(backpointer, prob=1.0)], prob=max_prob)
        elif isinstance(backpointer, tuple) and len(backpointer) == 1: # Unary non-terminal rule (e.g., ('N',))
            child_nt = backpointer[0]
            child_node = self._reconstruct_tree(child_nt, i, j, chart) # Unary rule doesn't change span (i,j)
            return ParseTreeNode(lhs, [child_node], prob=max_prob)
        else:  # Binary non-terminal rule (k, B, C)
            k, B, C = backpointer
            left_child = self._reconstruct_tree(B, i, k, chart)
            right_child = self._reconstruct_tree(C, k, j, chart)
            return ParseTreeNode(lhs, [left_child, right_child], prob=max_prob)

class ParseTreeNode:
    def __init__(self, label, children=None, prob=1.0):
        self.label = label
        self.children = children if children is not None else []
        self.prob = prob

    def __str__(self, level=0):
        ret = "  " * level + f"{self.label} (p={self.prob:.4f})\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.prob}, {self.children})"

# --- Define a sample PCFG ---
# Grammar rules are (probability, production_tuple)
# Note: For simplicity, all word productions (e.g., N -> cat) are assumed to have a probability of 1.0
# when they match directly from the lexicon. A more robust PCFG would include these probabilities.

pcfg_grammar = {
    'S': [
        (1.0, ('NP', 'VP'))
    ],
    'NP': [
        (0.6, ('Det', 'N')),
        (0.4, ('N'))
    ],
    'VP': [
        (0.7, ('V', 'NP')),
        (0.3, ('V'))
    ],
    'Det': [
        (0.6, ('the')),
        (0.4, ('a'))
    ],
    'N': [
        (0.5, ('cat')),
        (0.3, ('dog')),
        (0.2, ('man'))
    ],
    'V': [
        (0.5, ('chased')),
        (0.5, ('ate'))
    ]
}

# Test sentences
sentences_for_pcfg = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog"
]

pcfg_parser = PCFGParser(pcfg_grammar)

print("PCFG Parsing Results (Most Probable Parse Tree):\n")
for sentence in sentences_for_pcfg:
    print(f"Sentence: '{sentence}'")
    max_prob, parse_tree = pcfg_parser.parse(sentence)
    if parse_tree:
        print(f"  Accepted with probability: {max_prob:.6f}")
        print("  Parse Tree:")
        print(parse_tree)
    else:
        print("  Rejected. No parse found.")
```

18. Implement a simple FOPC parser for basic logical expressions using python program.

### Simple FOPC Parser in Python

This program implements a basic parser for First-Order Predicate Calculus (FOPC) expressions. It handles:

*   **Constants**: `a`, `b`, `c`, etc.
*   **Variables**: `x`, `y`, `z`, etc.
*   **Predicates**: `P(x)`, `Q(a,y)`, `R(c)`, etc.
*   **Logical Connectives**: `AND` (`&`), `OR` (`|`), `NOT` (`~`), `IMPLIES` (`->`), `BICONDITIONAL` (`<->`)
*   **Quantifiers**: `FORALL` (`FORALL x`), `EXISTS` (`EXISTS y`)

The parser first tokenizes the input string and then uses a recursive descent approach to build an Abstract Syntax Tree (AST) representing the logical expression.

```python
import re

# --- Tokenization ---

class Token:
    def __init__(self, type, value=None):
        self.type = type
        self.value = value

    def __repr__(self):
        return f"Token({self.type}, {self.value if self.value is not None else ''})"

class Lexer:
    def __init__(self, text):
        self.text = text
        self.pos = 0
        self.current_char = self.text[self.pos] if self.text else None

    def advance(self):
        self.pos += 1
        if self.pos < len(self.text):
            self.current_char = self.text[self.pos]
        else:
            self.current_char = None

    def skip_whitespace(self):
        while self.current_char is not None and self.current_char.isspace():
            self.advance()

    def get_next_token(self):
        while self.current_char is not None:
            if self.current_char.isspace():
                self.skip_whitespace()
                continue

            # Operators and Delimiters
            if self.current_char == '(':
                self.advance()
                return Token('LPAREN')
            if self.current_char == ')':
                self.advance()
                return Token('RPAREN')
            if self.current_char == '&':
                self.advance()
                return Token('AND')
            if self.current_char == '|':
                self.advance()
                return Token('OR')
            if self.current_char == '~':
                self.advance()
                return Token('NOT')
            if self.current_char == ',': # Add support for comma
                self.advance()
                return Token('COMMA')
            if self.current_char == '-' and self.peek() == '>':
                self.advance()
                self.advance()
                return Token('IMPLIES')
            if self.current_char == '<' and self.peek() == '-' and self.peek(2) == '>':
                self.advance()
                self.advance()
                self.advance()
                return Token('BICONDITIONAL')

            # Identifiers (Predicates, Variables, Constants, Quantifiers)
            if self.current_char.isalpha():
                start_pos = self.pos
                while self.current_char is not None and (self.current_char.isalnum() or self.current_char == '_'):
                    self.advance()
                identifier = self.text[start_pos:self.pos].upper()

                if identifier == 'FORALL':
                    return Token('FORALL')
                if identifier == 'EXISTS':
                    return Token('EXISTS')

                # Simple heuristic: lowercase single letters are variables/constants, uppercase are predicates
                # This is a simplification; a real FOPC would have explicit declarations or more complex rules.
                if len(identifier) == 1 and identifier.islower(): # Assume single lowercase char is a variable/constant
                    return Token('ID', identifier.lower())
                elif identifier.isupper() or len(identifier) > 1: # Assume other identifiers are predicates
                     return Token('ID', identifier)

            # If we reached here, it's an unrecognized character
            raise Exception(f"Lexer error: Unrecognized character '{self.current_char}' at position {self.pos}")

        return Token('EOF')

    def peek(self, offset=1):
        peek_pos = self.pos + offset
        if peek_pos < len(self.text):
            return self.text[peek_pos]
        return None

# --- AST Nodes ---

class Node:
    pass

class ConstantNode(Node):
    def __init__(self, value):
        self.value = value
    def __repr__(self):
        return f"Constant({self.value})"

class VariableNode(Node):
    def __init__(self, name):
        self.name = name
    def __repr__(self):
        return f"Variable({self.name})"

class PredicateNode(Node):
    def __init__(self, name, terms):
        self.name = name
        self.terms = terms # list of ConstantNode or VariableNode
    def __repr__(self):
        return f"Predicate({self.name}, {self.terms})"

class UnaryOpNode(Node):
    def __init__(self, op_token, operand):
        self.op = op_token
        self.operand = operand
    def __repr__(self):
        return f"UnaryOp({self.op.type}, {self.operand})"

class BinaryOpNode(Node):
    def __init__(self, left, op_token, right):
        self.left = left
        self.op = op_token
        self.right = right
    def __repr__(self):
        return f"BinaryOp({self.left}, {self.op.type}, {self.right})"

class QuantifierNode(Node):
    def __init__(self, quantifier_token, variable_token, expression):
        self.quantifier = quantifier_token
        self.variable = variable_token
        self.expression = expression
    def __repr__(self):
        return f"Quantifier({self.quantifier.type} {self.variable.value}, {self.expression})"

# --- Parser ---

class Parser:
    def __init__(self, lexer):
        self.lexer = lexer
        self.current_token = self.lexer.get_next_token()

    def error(self, message="Invalid syntax"):
        raise Exception(f"Parser error: {message} at token {self.current_token}")

    def eat(self, token_type):
        if self.current_token.type == token_type:
            self.current_token = self.lexer.get_next_token()
        else:
            self.error(f"Expected {token_type}, got {self.current_token.type}")

    def term(self):
        token = self.current_token
        if token.type == 'ID':
            # Check if it's a predicate (followed by LPAREN)
            next_token = self.lexer.peek_token()
            if next_token and next_token.type == 'LPAREN':
                self.eat('ID') # Eat predicate name
                predicate_name = token.value
                self.eat('LPAREN')
                terms = []
                while self.current_token.type != 'RPAREN':
                    term_id = self.current_token
                    self.eat('ID') # Eat variable/constant for term
                    # Simplified: assume ID tokens in terms are Variables
                    terms.append(VariableNode(term_id.value))
                    if self.current_token.type == 'COMMA':
                        self.eat('COMMA')
                self.eat('RPAREN')
                return PredicateNode(predicate_name, terms)
            else:
                self.eat('ID')
                # Simplified: assume single letter IDs are variables/constants
                if len(token.value) == 1 and token.value.islower():
                    return VariableNode(token.value)
                else:
                    return ConstantNode(token.value) # Fallback for multi-char IDs without parens
        else:
            self.error("Expected a term (predicate, variable, or constant)")

    def primary_expression(self):
        token = self.current_token
        if token.type == 'NOT':
            self.eat('NOT')
            return UnaryOpNode(token, self.primary_expression())
        elif token.type == 'FORALL' or token.type == 'EXISTS':
            quantifier_token = token
            self.eat(token.type)
            variable_token = self.current_token
            self.eat('ID') # Variable for quantification
            expression = self.primary_expression() # The scope of the quantifier
            return QuantifierNode(quantifier_token, variable_token, expression)
        elif token.type == 'LPAREN':
            self.eat('LPAREN')
            node = self.expression()
            self.eat('RPAREN')
            return node
        else:
            return self.term()

    def binary_operation(self, left_node, op_types, next_precedence_func):
        while self.current_token.type in op_types:
            op_token = self.current_token
            self.eat(op_token.type)
            right_node = next_precedence_func()
            left_node = BinaryOpNode(left_node, op_token, right_node)
        return left_node

    def biconditional_expression(self):
        node = self.implication_expression()
        return self.binary_operation(node, ['BICONDITIONAL'], self.implication_expression)

    def implication_expression(self):
        node = self.or_expression()
        return self.binary_operation(node, ['IMPLIES'], self.or_expression)

    def or_expression(self):
        node = self.and_expression()
        return self.binary_operation(node, ['OR'], self.and_expression)

    def and_expression(self):
        node = self.primary_expression()
        return self.binary_operation(node, ['AND'], self.primary_expression)

    def expression(self):
        return self.biconditional_expression()

    def parse(self):
        node = self.expression()
        if self.current_token.type != 'EOF':
            self.error("Unexpected tokens at end of expression")
        return node

# Helper for Lexer to peek ahead without consuming
def _peek_token_in_lexer(self):
    original_pos = self.pos
    original_char = self.current_char
    # Temporarily store the original text to restore it later if a full reset is needed
    # This is a bit of a hack around how get_next_token advances state.
    # A better lexer would have a way to 'unget' tokens or truly peek without state change.
    # For this simple example, resetting pos/char is sufficient.
    saved_text_view = self.text[self.pos:] # Save the remaining text

    token = self.get_next_token()

    # Restore original state. This might not perfectly restore `current_char`
    # if `get_next_token` internally skips whitespace and changes `current_char` multiple times.
    # For robustness, we should re-initialize from `original_pos`.
    self.pos = original_pos
    self.current_char = self.text[self.pos] if self.pos < len(self.text) else None

    return token

Lexer.peek_token = _peek_token_in_lexer # Add peek_token method to Lexer dynamically

# --- Test Cases ---

print("FOPC Parser Test Cases:")

test_expressions = [
    "P(x) & Q(a)",
    "~R(y) | S(b,z)",
    "FORALL x (P(x) -> Q(x))",
    "EXISTS y (R(y) & ~S(y))",
    "A <-> B",
    "P(x) & (Q(y) | R(z))",
    "FORALL x EXISTS y P(x, y)" # Nested quantifiers
]

for expr_str in test_expressions:
    print(f"\nParsing: '{expr_str}'")
    try:
        lexer = Lexer(expr_str)
        parser = Parser(lexer)
        ast = parser.parse()
        print("  AST:", ast)
    except Exception as e:
        print("  Error:", e)
```

20. Implement a basic information retrieval system using TF-IDF (Term Frequency-Inverse
Document Frequency) for document ranking using python.

### Information Retrieval with TF-IDF

TF-IDF (Term Frequency-Inverse Document Frequency) is a numerical statistic that reflects how important a word is to a document in a collection or corpus. It is often used as a weighting factor in information retrieval and text mining. The TF-IDF value increases proportionally to the number of times a word appears in the document but is offset by the frequency of the word in the corpus, which helps to adjust for the fact that some words appear more frequently in general.

This implementation will:
1.  Define a collection of sample documents.
2.  Compute the TF-IDF matrix for these documents.
3.  Calculate the cosine similarity between a given query and each document.
4.  Rank the documents based on their similarity scores.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Define a sample corpus of documents
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Never jump over the lazy dog with a fox.",
    "The dog is lazy and brown.",
    "A quick fox is a brown fox.",
    "The cat sat on the mat."
]

print("--- Documents ---")
for i, doc in enumerate(documents):
    print(f"Document {i+1}: {doc}")
print("\n")
```

```python
# 2. Calculate TF-IDF scores

# Initialize TfidfVectorizer
# It will tokenize the documents, build a vocabulary, and compute TF-IDF scores
vectorizer = TfidfVectorizer()

# Fit the vectorizer to the documents and transform them into a TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

print("--- TF-IDF Matrix Shape ---")
print(tfidf_matrix.shape) # (number of documents, number of unique words)

# You can inspect the feature names (words in the vocabulary)
print("\n--- Feature Names (Vocabulary) ---")
print(vectorizer.get_feature_names_out())

# To see the actual matrix (sparse format), convert to dense array
# print("\n--- TF-IDF Matrix (Dense) ---")
# print(tfidf_matrix.toarray())
```

```python
# 3. Implement a function to rank documents based on a query

def rank_documents(query, documents, vectorizer, tfidf_matrix):
    # Transform the query into a TF-IDF vector using the same vectorizer
    query_vector = vectorizer.transform([query])

    # Calculate cosine similarity between the query vector and all document vectors
    # cosine_similarity returns a matrix, we need the first row (similarities with the query)
    similarities = cosine_similarity(query_vector, tfidf_matrix).flatten()

    # Get the indices of documents sorted by similarity in descending order
    ranked_indices = similarities.argsort()[::-1]

    # Prepare the ranked results
    ranked_results = []
    for idx in ranked_indices:
        ranked_results.append({
            'document_id': idx + 1,
            'document_text': documents[idx],
            'similarity_score': similarities[idx]
        })

    return ranked_results

print("Document ranking function 'rank_documents' defined.")
```

```python
# 4. Demonstrate the system with a sample query

sample_query = "lazy dog and fox"
print(f"\n--- Ranking Documents for Query: '{sample_query}' ---")

results = rank_documents(sample_query, documents, vectorizer, tfidf_matrix)

for result in results:
    print(f"Document {result['document_id']} (Score: {result['similarity_score']:.4f}): {result['document_text']}")
```

```python
# Try another query
sample_query_2 = "quick brown cat"
print(f"\n--- Ranking Documents for Query: '{sample_query_2}' ---")

results_2 = rank_documents(sample_query_2, documents, vectorizer, tfidf_matrix)

for result in results_2:
    print(f"Document {result['document_id']} (Score: {result['similarity_score']:.4f}): {result['document_text']}")
```

### Syntax-Driven Semantic Analysis: Noun Phrase Extraction and Meaning

This program demonstrates a basic form of syntax-driven semantic analysis. It involves:

1.  **Tokenization and Part-of-Speech (POS) Tagging**: Breaking the sentence into words and assigning a grammatical category (e.g., noun, verb, adjective) to each word.
2.  **Noun Phrase (NP) Chunking**: Identifying sequences of words that form noun phrases based on grammatical rules.
3.  **Semantic Interpretation**: For each extracted noun phrase, identifying its head noun and retrieving its definition(s) from WordNet as a basic representation of its meaning.

```python
import nltk
from nltk.corpus import wordnet
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

# Download necessary NLTK data if not already present
# 'punkt' for tokenization, 'averaged_perceptron_tagger' for POS tagging, 'wordnet' for meanings.
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    print("Downloading 'punkt' tokenizer...")
    nltk.download('punkt')
try:
    nltk.data.find('taggers/averaged_perceptron_tagger')
except LookupError:
    print("Downloading 'averaged_perceptron_tagger'...")
    nltk.download('averaged_perceptron_tagger')
try:
    nltk.data.find('taggers/averaged_perceptron_tagger_eng') # Added download for this specific resource
except LookupError:
    print("Downloading 'averaged_perceptron_tagger_eng'...")
    nltk.download('averaged_perceptron_tagger_eng')
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    print("Downloading 'wordnet' corpus...")
    nltk.download('wordnet')


def extract_noun_phrases_and_meanings(sentence):
    # 1. Tokenize and POS tag the sentence
    tokens = word_tokenize(sentence)
    tagged_words = pos_tag(tokens)
    print(f"\nTagged words: {tagged_words}")

    # 2. Define a grammar for Noun Phrases (NP)
    # Example: NP: {<DT>?<JJ>*<NN>+}
    # DT (Determiner), JJ (Adjective), NN (Noun)
    # This rule captures optional determiners, zero or more adjectives, followed by one or more nouns.
    grammar = r"""
      NP: {<DT|JJ|NN.*>+}
    """
    # NP: {<DT>?<JJ>*<NN.*>+}

    cp = nltk.RegexpParser(grammar)
    tree = cp.parse(tagged_words)
    # print("\nParse Tree:")
    # tree.pretty_print()

    noun_phrases = []
    for subtree in tree.subtrees():
        if subtree.label() == 'NP':
            # Extract words from the NP subtree
            np_words = [word for word, tag in subtree.leaves()]
            noun_phrases.append(np_words)

    print(f"\nExtracted Noun Phrases: {noun_phrases}")

    # 3. Extract meanings for head nouns in each NP
    print("\n--- Semantic Analysis (Meanings) ---")
    results = {}
    for np in noun_phrases:
        # Simple heuristic: assume the last noun in the NP is the head noun
        head_noun = None
        for word_in_np in reversed(np):
            # Check if the word is a noun (NN, NNS, NNP, NNPS)
            for w, t in tagged_words:
                if w == word_in_np and t.startswith('NN'):
                    head_noun = word_in_np
                    break
            if head_noun:
                break

        if head_noun:
            print(f"\nNoun Phrase: '{' '.join(np)}' (Head Noun: '{head_noun}')")
            synsets = wordnet.synsets(head_noun)
            if synsets:
                results[' '.join(np)] = []
                print(f"  Meanings for '{head_noun}':")
                for i, synset in enumerate(synsets[:3]): # Limit to first 3 for brevity
                    print(f"    {i+1}. {synset.definition()}")
                    results[' '.join(np)].append(synset.definition())
            else:
                print(f"  No meanings found for '{head_noun}' in WordNet.")
        else:
            print(f"\nNoun Phrase: '{' '.join(np)}' (No discernible head noun for meaning lookup)")

    return results

# --- Test Cases ---
sentence1 = "The quick brown fox jumps over the lazy dog."
sentence2 = "Artificial intelligence is a rapidly growing field."
sentence3 = "A big red ball rolled down the steep hill."

print(f"Parsing Sentence 1: '{sentence1}'")
extract_noun_phrases_and_meanings(sentence1)

print(f"\n{'='*50}\n")

print(f"Parsing Sentence 2: '{sentence2}'")
extract_noun_phrases_and_meanings(sentence2)

print(f"\n{'='*50}\n")

print(f"Parsing Sentence 3: '{sentence3}'")
extract_noun_phrases_and_meanings(sentence3)
```

### Knowledge Graph Extraction (Subject-Verb-Object Triples)

This program demonstrates how to extract Subject-Verb-Object (SVO) triples from sentences to build a rudimentary knowledge graph. It leverages SpaCy's advanced linguistic processing capabilities, particularly its dependency parser, to identify the grammatical relationships between words.

**Steps:**
1.  **Load SpaCy Model**: Initialize an English language model.
2.  **Sentence Processing**: Process a text to obtain a document object with tokens, POS tags, and dependency parses.
3.  **Triple Extraction**: Iterate through sentences and apply rules based on dependency relations to find subjects, verbs, and objects.

```python
import spacy

# Load the English language model. Download it if not already present.
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading 'en_core_web_sm' model...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

def extract_svo_triples(text):
    doc = nlp(text)
    svo_triples = []

    for sent in doc.sents:
        subject = ""
        verb = ""
        obj = ""

        # Find the main verb in the sentence
        # Often, the root of the dependency tree for a sentence is a verb
        for token in sent:
            if token.dep_ == "ROOT" and token.pos_ == "VERB":
                verb = token
                break

        if not verb:
            continue # Skip if no main verb found

        # Find the subject(s) of the verb
        subjects = [w for w in sent if w.dep_ in ("nsubj", "nsubjpass") and w.head == verb]
        if subjects:
            subject = " ".join([w.text for w in subjects])

        # Find the direct object(s) or object of a preposition of the verb
        objects = [w for w in sent if w.dep_ in ("dobj", "pobj") and w.head == verb]
        if not objects:
            # Try to find objects connected via compounds
            objects = [w for w in sent if w.dep_ == "attr" and w.head == verb]

        if objects:
            obj = " ".join([w.text for w in objects])

        # Simple heuristic to get full phrase for subject/object if they are noun chunks
        # This makes the output more readable
        for chunk in sent.noun_chunks:
            if subject and verb and chunk.root.text in subject and chunk.root.head == verb:
                subject = chunk.text
            if obj and verb and chunk.root.text in obj and chunk.root.head == verb:
                obj = chunk.text

        # Store the triple if a subject, verb, and (optionally) an object are found
        if subject and verb:
            triple = {
                "subject": subject,
                "verb": verb.lemma_,
                "object": obj if obj else None
            }
            svo_triples.append(triple)

    return svo_triples

# --- Test Text ---
text1 = "Barack Obama was born in Hawaii. He served as the 44th President of the United States. He is married to Michelle Obama."
text2 = "The cat sat on the mat. It chased a mouse. A big dog barked loudly."
text3 = "Apple released a new iPhone. Consumers love its features. Tim Cook is the CEO."

print("--- Knowledge Graph Extraction ---")

print(f"\nProcessing text:\n'{text1}'")
triples1 = extract_svo_triples(text1)
for triple in triples1:
    print(f"  Subject: {triple['subject']}, Verb: {triple['verb']}, Object: {triple['object']}")

print(f"\nProcessing text:\n'{text2}'")
triples2 = extract_svo_triples(text2)
for triple in triples2:
    print(f"  Subject: {triple['subject']}, Verb: {triple['verb']}, Object: {triple['object']}")

print(f"\nProcessing text:\n'{text3}'")
triples3 = extract_svo_triples(text3)
for triple in triples3:
    print(f"  Subject: {triple['subject']}, Verb: {triple['verb']}, Object: {triple['object']}")
```

23. Develop a python program that evaluates the coherence of a given text.

### Text Coherence Evaluation using Sentence Embeddings

This program evaluates the coherence of a given text by measuring the semantic similarity between consecutive sentences. It uses a pre-trained `SentenceTransformer` model to generate embeddings for each sentence and then calculates the cosine similarity between these embeddings.

A higher average cosine similarity between adjacent sentences generally indicates a more coherent text, as the semantic content flows more smoothly from one sentence to the next.

```python
# Install the sentence-transformers library if you haven't already
!pip install sentence-transformers

import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import nltk
from nltk.tokenize import sent_tokenize

# Download NLTK punkt tokenizer for sentence splitting
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    print("Downloading 'punkt' tokenizer...")
    nltk.download('punkt')
```

```python
def evaluate_text_coherence(text, model_name='all-MiniLM-L6-v2'):
    """
    Evaluates the coherence of a text by calculating the average cosine similarity
    between adjacent sentence embeddings.

    Args:
        text (str): The input text to evaluate.
        model_name (str): The name of the SentenceTransformer model to use.

    Returns:
        tuple: A tuple containing:
            - float: The average coherence score (average cosine similarity).
            - list: A list of individual cosine similarity scores between adjacent sentences.
    """
    # Load a pre-trained sentence transformer model
    model = SentenceTransformer(model_name)

    # Split the text into sentences
    sentences = sent_tokenize(text)

    if len(sentences) < 2:
        return 1.0, [] # A single sentence is considered perfectly coherent with itself

    # Generate embeddings for each sentence
    sentence_embeddings = model.encode(sentences)

    # Calculate cosine similarity between adjacent sentence embeddings
    coherence_scores = []
    for i in range(len(sentence_embeddings) - 1):
        sim = cosine_similarity(
            sentence_embeddings[i].reshape(1, -1),
            sentence_embeddings[i+1].reshape(1, -1)
        )[0][0]
        coherence_scores.append(sim)

    # Calculate the average coherence score
    average_coherence = np.mean(coherence_scores)

    return average_coherence, coherence_scores
```

```python
# --- Test Cases ---

# Example 1: Highly coherent text
coherent_text = (
    "The sun rises in the east. It brings warmth and light to the world. "
    "Birds start chirping, signaling the start of a new day. People wake up and begin their routines."
)

print("\n--- Coherent Text Example ---")
print(f"Text:\n{coherent_text}")
avg_coh_1, scores_1 = evaluate_text_coherence(coherent_text)
print(f"Average Coherence Score: {avg_coh_1:.4f}")
print(f"Individual Scores: {scores_1}")

# Example 2: Less coherent text (topic shifts)
less_coherent_text = (
    "I enjoy hiking in the mountains. My favorite color is blue. "
    "The capital of France is Paris. Cats are mammals with sharp claws."
)

print("\n--- Less Coherent Text Example ---")
print(f"Text:\n{less_coherent_text}")
avg_coh_2, scores_2 = evaluate_text_coherence(less_coherent_text)
print(f"Average Coherence Score: {avg_coh_2:.4f}")
print(f"Individual Scores: {scores_2}")

# Example 3: Moderately coherent text
moderately_coherent_text = (
    "Artificial intelligence is transforming many industries. Machine learning is a key component of AI. "
    "However, ethical considerations are also crucial. Developing robust and fair AI systems is a significant challenge."
)

print("\n--- Moderately Coherent Text Example ---")
print(f"Text:\n{moderately_coherent_text}")
avg_coh_3, scores_3 = evaluate_text_coherence(moderately_coherent_text)
print(f"Average Coherence Score: {avg_coh_3:.4f}")
print(f"Individual Scores: {scores_3}")
```

24. Create a python program that recognizes dialog acts in a given dialog or conversation

### Dialog Act Recognition

Dialog act recognition involves classifying utterances in a conversation based on their communicative function (e.g., question, statement, command, acknowledgement). This is a fundamental task in natural language understanding, especially for conversational AI systems.

This program demonstrates a basic approach using:
1.  **Feature Extraction**: Converting text utterances into numerical features (e.g., using TF-IDF).
2.  **Classification Model**: Training a machine learning model (like Logistic Regression) to map these features to predefined dialog acts.

**Note**: For production-level dialog act recognition, more sophisticated models, richer features, and large, annotated datasets are typically required.

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import numpy as np

# Download necessary NLTK data if not already present
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    print("Downloading 'punkt' tokenizer...")
    nltk.download('punkt')

# --- Mock Training Data ---
# In a real-world scenario, you would use a much larger and more diverse dataset
# labeled by human annotators.

training_data = [
    ("Hi there!", "greeting"),
    ("Hello.", "greeting"),
    ("How are you doing today?", "question"),
    ("I'm doing great, thanks!", "statement"),
    ("Could you please tell me your name?", "question"),
    ("My name is John Doe.", "statement"),
    ("What is the capital of France?", "question"),
    ("Paris is the capital of France.", "statement"),
    ("Thank you for the information.", "thanks"),
    ("You're welcome!", "acknowledgement"),
    ("Please close the door.", "command"),
    ("Okay, I will.", "acknowledgement"),
    ("I agree with that.", "agreement"),
    ("I disagree entirely.", "disagreement"),
    ("Tell me about the weather.", "command"),
    ("The weather is sunny today.", "statement"),
    ("Are you sure?", "question"),
    ("Yes, I am sure.", "statement"),
    ("Goodbye!", "farewell"),
    ("See you later!", "farewell")
]

# Separate utterances and labels
utterances = [item[0] for item in training_data]
labels = [item[1] for item in training_data]

# --- Build and Train the Model ---
# We'll use a simple pipeline: TF-IDF for feature extraction and Logistic Regression for classification.

# Split data into training and testing sets (for evaluation purposes)
X_train, X_test, y_train, y_test = train_test_split(utterances, labels, test_size=0.2, random_state=42)

# Create a pipeline
text_clf = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression(random_state=42))
])

# Train the classifier
print("Training the dialog act recognition model...")
text_clf.fit(X_train, y_train)
print("Model training complete.")

# Evaluate the model (optional, but good practice)
y_pred = text_clf.predict(X_test)
print("\n--- Model Evaluation Report ---")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
def recognize_dialog_acts(conversation_text, model):
    """
    Recognizes dialog acts in a given conversation text.

    Args:
        conversation_text (str): The input conversation text.
        model: The trained scikit-learn pipeline for dialog act classification.

    Returns:
        list: A list of dictionaries, where each dictionary contains the
              utterance and its predicted dialog act.
    """
    sentences = sent_tokenize(conversation_text)
    predictions = []
    for sentence in sentences:
        # Predict the dialog act for each sentence
        predicted_act = model.predict([sentence])[0]
        predictions.append({"utterance": sentence.strip(), "dialog_act": predicted_act})
    return predictions

# --- Demonstrate with a Sample Conversation ---

sample_conversation = (
    "Hi there! How can I help you today? I have a question about my order. "
    "Could you please tell me the order number? It is 12345. Thank you. "
    "You're welcome! Is there anything else? No, that's all. Goodbye!"
)

print("\n--- Sample Conversation ---")
print(sample_conversation)

print("\n--- Predicted Dialog Acts ---")
predicted_acts = recognize_dialog_acts(sample_conversation, text_clf)
for item in predicted_acts:
    print(f"Utterance: '{item['utterance']}' -> Dialog Act: {item['dialog_act']}")


print("\n--- Another Example ---")
another_conversation = (
    "Hello. What is your opinion on AI? I think it's revolutionary. "
    "Do you agree? Yes, I agree completely. Please elaborate. "
    "It will change the world. See you later!"
)

print(another_conversation)
print("\n--- Predicted Dialog Acts ---")
predicted_acts_2 = recognize_dialog_acts(another_conversation, text_clf)
for item in predicted_acts_2:
    print(f"Utterance: '{item['utterance']}' -> Dialog Act: {item['dialog_act']}")
```

### 4. Generate Text based on a Prompt

Now you can use the client to send a request to the GPT model. We'll use the `chat.completions.create` endpoint, which is the standard for interacting with GPT-3.5 Turbo and GPT-4 models.

You provide a list of messages, where each message has a `role` (e.g., 'user', 'system', 'assistant') and `content`. The model then generates a response based on this conversation history.

```python
def generate_text_with_gpt(prompt_text, model_name="gpt-3.5-turbo", max_tokens=150, temperature=0.7):
    """
    Generates text using the OpenAI GPT model based on a given prompt.

    Args:
        prompt_text (str): The input prompt for text generation.
        model_name (str): The name of the GPT model to use (e.g., 'gpt-3.5-turbo').
        max_tokens (int): The maximum number of tokens to generate in the response.
        temperature (float): Controls the randomness of the output. Higher values mean more random.

    Returns:
        str: The generated text.
    """
    try:
        response = client.chat.completions.create(
            model=model_name,
            messages=[
                {"role": "system", "content": "You are a helpful assistant."}, # Optional system message to set context
                {"role": "user", "content": prompt_text}
            ],
            max_tokens=max_tokens,
            temperature=temperature
        )
        return response.choices[0].message.content.strip()
    except openai.APIError as e:
        print(f"OpenAI API Error: {e}")
        return "Error: Could not generate text."
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return "Error: Could not generate text."

# --- Test Cases ---

# Prompt 1: Simple creative writing
prompt1 = "Write a short, whimsical story about a cat who learns to fly."
print("\n--- Generating Text for Prompt 1 ---")
print(f"Prompt: '{prompt1}'")
generated_story = generate_text_with_gpt(prompt1)
print("Generated Text:\n")
print(generated_story)

# Prompt 2: Factual query
prompt2 = "Explain the concept of quantum entanglement in simple terms."
print("\n--- Generating Text for Prompt 2 ---")
print(f"Prompt: '{prompt2}'")
generated_explanation = generate_text_with_gpt(prompt2, max_tokens=200, temperature=0.5)
print("Generated Text:\n")
print(generated_explanation)

# Prompt 3: Code generation (example)
prompt3 = "Write a Python function to reverse a string."
print("\n--- Generating Text for Prompt 3 ---")
print(f"Prompt: '{prompt3}'")
generated_code = generate_text_with_gpt(prompt3, max_tokens=100, temperature=0.2)
print("Generated Text:\n")
print(generated_code)
```

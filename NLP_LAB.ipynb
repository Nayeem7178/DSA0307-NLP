EXP 02- Implement a basic finite state automaton that recognizes a specific language or pattern. In this
example, we'll create a simple automaton to match strings ending with 'ab' using python


```python
def fsa(string):
    state = 0

    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
            else:
                state = 0
        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

s = input("Enter a string: ")
fsa(s)
```

```python
import random

text = "the cat sat on the mat the cat ate the fish"

words = text.split()

bigrams = {}

for i in range(len(words) - 1):
    key = words[i]
    value = words[i + 1]
    if key not in bigrams:
        bigrams[key] = []
    bigrams[key].append(value)

word = random.choice(words)
result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print("Generated Text:")
print(" ".join(result))
```

11. Implement a simple top-down parser for context-free grammars using python.

### Top-Down Parser for Context-Free Grammars

This program implements a basic top-down parser. It attempts to derive a given input sentence from a defined context-free grammar by recursively expanding non-terminal symbols until terminals match the input.

```python
class Parser:
    def __init__(self, grammar):
        self.grammar = grammar
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0
        return self.parse_symbol(self.grammar['S'])

    def parse_symbol(self, productions):
        original_index = self.current_token_index
        for production in productions:
            self.current_token_index = original_index
            if self.match_production(production):
                return True
        return False

    def match_production(self, production):
        for symbol in production:
            if symbol in self.grammar:  # Non-terminal
                if not self.parse_symbol(self.grammar[symbol]):
                    return False
            else:  # Terminal
                if not self.match_terminal(symbol):
                    return False
        return True

    def match_terminal(self, terminal):
        if self.current_token_index < len(self.tokens) and self.tokens[self.current_token_index] == terminal:
            self.current_token_index += 1
            return True
        return False

# Define a simple context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test cases
sentences = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog",
    "chased the dog"
]

parser = Parser(grammar)

print("Parsing Results:")
for sentence in sentences:
    if parser.parse(sentence) and parser.current_token_index == len(parser.tokens):
        print(f"'{sentence}' -> Accepted")
    else:
        print(f"'{sentence}' -> Rejected")
```

12. Implement an Earley parser for context-free grammars using a simple python program.

### Earley Parser for Context-Free Grammars

An Earley parser is a chart parser that uses dynamic programming to parse sentences according to a context-free grammar. It's known for its ability to handle all context-free grammars, including ambiguous and left-recursive ones.

```python
class EarleyParser:
    def __init__(self, grammar, start_symbol='S'):
        self.grammar = grammar
        self.start_symbol = start_symbol
        self.chart = []

    def parse(self, words):
        n = len(words)
        self.chart = [[] for _ in range(n + 1)]

        # Initialize chart[0] with the start symbol
        for prod in self.grammar.get(self.start_symbol, []):
            self.chart[0].append((self.start_symbol, prod, 0, 0))

        for i in range(n + 1):
            for item in self.chart[i]:
                rule_lhs, rule_rhs, dot_pos, origin = item

                if dot_pos < len(rule_rhs):  # Not a complete item
                    next_symbol = rule_rhs[dot_pos]
                    if next_symbol in self.grammar:  # Non-terminal
                        self.predict(next_symbol, i)
                    else:  # Terminal
                        self.scan(item, words, i)
                else:  # Complete item
                    self.complete(item, i)

        # Check if the sentence is accepted
        for item in self.chart[n]:
            if item[0] == self.start_symbol and item[2] == len(item[1]) and item[3] == 0:
                return True
        return False

    def predict(self, non_terminal, j):
        for prod in self.grammar.get(non_terminal, []):
            new_item = (non_terminal, prod, 0, j)
            if new_item not in self.chart[j]:
                self.chart[j].append(new_item)

    def scan(self, item, words, j):
        rule_lhs, rule_rhs, dot_pos, origin = item
        next_symbol = rule_rhs[dot_pos]
        if j < len(words) and words[j] == next_symbol:
            new_item = (rule_lhs, rule_rhs, dot_pos + 1, origin)
            if new_item not in self.chart[j + 1]:
                self.chart[j + 1].append(new_item)

    def complete(self, item, j):
        rule_lhs, rule_rhs, dot_pos, origin = item
        for prev_item in self.chart[origin]:
            prev_lhs, prev_rhs, prev_dot_pos, prev_origin = prev_item
            if prev_dot_pos < len(prev_rhs) and prev_rhs[prev_dot_pos] == rule_lhs:
                new_item = (prev_lhs, prev_rhs, prev_dot_pos + 1, prev_origin)
                if new_item not in self.chart[j]:
                    self.chart[j].append(new_item)

# Define a simple context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test cases
sentences = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog",
    "chased the dog",
    "a cat"
]

earley_parser = EarleyParser(grammar)

print("Earley Parser Results:")
for sentence in sentences:
    words = sentence.split()
    if earley_parser.parse(words):
        print(f"'{sentence}' -> Accepted")
    else:
        print(f"'{sentence}' -> Rejected")
```

13. Generate a parse tree for a given sentence using a context-free grammar using python program.

### Generating a Parse Tree from a Context-Free Grammar

This program implements a recursive descent parser with backtracking to construct a parse tree for a given sentence based on a context-free grammar. The `ParseTreeNode` class helps in representing the tree structure.

```python
class ParseTreeNode:
    def __init__(self, label, children=None):
        self.label = label
        self.children = children if children is not None else []

    def __str__(self, level=0):
        ret = "  " * level + self.label + "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.children})"


class RecursiveDescentParserWithTree:
    def __init__(self, grammar, start_symbol='S'):
        self.grammar = grammar
        self.start_symbol = start_symbol
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0

        root_node = self._parse_non_terminal(self.start_symbol)

        if root_node and self.current_token_index == len(self.tokens):
            return root_node
        else:
            return None

    def _parse_non_terminal(self, non_terminal_symbol):
        original_token_index = self.current_token_index

        for production in self.grammar.get(non_terminal_symbol, []):
            self.current_token_index = original_token_index # Backtrack
            children_nodes = []
            match_success = True

            for symbol in production:
                if symbol in self.grammar:  # Is a non-terminal
                    child_node = self._parse_non_terminal(symbol)
                    if child_node:
                        children_nodes.append(child_node)
                    else:
                        match_success = False
                        break
                else:  # Is a terminal
                    child_node = self._parse_terminal(symbol)
                    if child_node:
                        children_nodes.append(child_node)
                    else:
                        match_success = False
                        break

            if match_success:
                return ParseTreeNode(non_terminal_symbol, children_nodes)

        return None # No production matched

    def _parse_terminal(self, terminal_symbol):
        if self.current_token_index < len(self.tokens) and \
           self.tokens[self.current_token_index] == terminal_symbol:
            node = ParseTreeNode(terminal_symbol)
            self.current_token_index += 1
            return node
        return None

# Define the context-free grammar (same as before)
grammar = {
    'S': [['NP', 'VP']],
    'NP': [['Det', 'N'], ['N']],
    'VP': [['V', 'NP'], ['V']],
    'Det': [['the'], ['a']],
    'N': [['cat'], ['dog'], ['man']],
    'V': [['chased'], ['ate']]
}

# Test sentences
sentences_for_tree = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog" # This sentence will likely not form a complete 'S' tree
]

r_parser = RecursiveDescentParserWithTree(grammar)

print("Parse Tree Generation Results:")
for sentence in sentences_for_tree:
    print(f"\nSentence: '{sentence}'")
    parse_tree = r_parser.parse(sentence)
    if parse_tree:
        print("Accepted. Parse Tree:")
        print(parse_tree)
    else:
        print("Rejected. No complete parse tree found.")
```

14. Create a program in python to check for agreement in sentences based on a context-free
grammar's rules.

### Agreement Checking in Context-Free Grammars

This program implements a recursive descent parser capable of checking for feature agreement (e.g., subject-verb number agreement). The grammar and lexicon are extended to include features for symbols, and the parser enforces these constraints during parsing.

```python
class ParseTreeNode:
    def __init__(self, label, children=None, features=None):
        self.label = label
        self.children = children if children is not None else []
        self.features = features if features is not None else {}

    def __str__(self, level=0):
        ret = "  " * level + self.label
        if self.features:
            ret += f" {self.features}"
        ret += "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.features}, {self.children})"


class FeatureParser:
    def __init__(self, grammar, lexicon, start_symbol='S'):
        self.grammar = grammar
        self.lexicon = lexicon
        self.start_symbol = start_symbol
        self.tokens = []
        self.current_token_index = 0

    def parse(self, sentence):
        self.tokens = sentence.split()
        self.current_token_index = 0

        # Try to parse the start symbol
        result = self._parse_non_terminal(self.start_symbol)

        # A successful parse means we got a tree and consumed all tokens
        if result and self.current_token_index == len(self.tokens):
            return result[0]  # Return the ParseTreeNode
        else:
            return None

    def _parse_non_terminal(self, non_terminal_symbol):
        original_token_index = self.current_token_index

        for production_tuple in self.grammar.get(non_terminal_symbol, []):
            production_rule, agreement_func = production_tuple
            self.current_token_index = original_token_index  # Backtrack

            matched_children = []
            child_features = [] # To pass to agreement function
            match_success = True

            for symbol in production_rule:
                if symbol in self.grammar:  # It's a non-terminal
                    child_result = self._parse_non_terminal(symbol)
                    if child_result:
                        child_node, child_feat = child_result
                        matched_children.append(child_node)
                        child_features.append(child_feat)
                    else:
                        match_success = False
                        break
                else:  # It's a terminal (look up in lexicon)
                    child_result = self._parse_terminal(symbol)
                    if child_result:
                        child_node, child_feat = child_result
                        matched_children.append(child_node)
                        child_features.append(child_feat)
                    else:
                        match_success = False
                        break

            if match_success:
                # Apply agreement function if provided
                if agreement_func:
                    # Pass features of children to the agreement function
                    # The agreement function should return parent features or None if agreement fails
                    parent_features = agreement_func(*child_features)
                    if parent_features is None:
                        match_success = False # Agreement failed
                else:
                    parent_features = {}

            if match_success:
                return ParseTreeNode(non_terminal_symbol, matched_children, parent_features), parent_features

        return None

    def _parse_terminal(self, expected_pos_tag):
        # Check if current token matches any word with the expected POS tag
        if self.current_token_index < len(self.tokens):
            current_word = self.tokens[self.current_token_index]
            word_info = self.lexicon.get(current_word)

            if word_info and word_info['POS'] == expected_pos_tag:
                node = ParseTreeNode(current_word, features=word_info)
                self.current_token_index += 1
                return node, word_info
        return None

# --- Define Lexicon and Grammar with Agreement Rules ---

lexicon = {
    'the': {'POS': 'Det', 'NUM': 'any'},
    'a': {'POS': 'Det', 'NUM': 'singular'},
    'cat': {'POS': 'N', 'NUM': 'singular'},
    'cats': {'POS': 'N', 'NUM': 'plural'},
    'dog': {'POS': 'N', 'NUM': 'singular'},
    'dogs': {'POS': 'N', 'NUM': 'plural'},
    'man': {'POS': 'N', 'NUM': 'singular'},
    'men': {'POS': 'N', 'NUM': 'plural'},
    'runs': {'POS': 'V', 'NUM': 'singular'}, # V for singular subject
    'run': {'POS': 'V', 'NUM': 'plural'},   # V for plural subject
    'chases': {'POS': 'V', 'NUM': 'singular'},
    'chase': {'POS': 'V', 'NUM': 'plural'},
    'eats': {'POS': 'V', 'NUM': 'singular'},
    'eat': {'POS': 'V', 'NUM': 'plural'}
}

# Agreement functions:
def s_agreement_check(np_features, vp_features):
    # S inherits number from NP/VP if they agree
    if np_features['NUM'] == vp_features['NUM']:
        return {'NUM': np_features['NUM']}
    return None # Agreement failure

def np_det_n_agreement_check(det_features, n_features):
    # Det must agree with N, or be 'any' (like 'the')
    if det_features['NUM'] == 'any' or det_features['NUM'] == n_features['NUM']:
        return {'NUM': n_features['NUM']}
    return None

def np_n_agreement_check(n_features):
    # NP inherits number from N
    return {'NUM': n_features['NUM']}

def vp_v_np_agreement_check(v_features, np_features):
    # VP inherits number from V (V-O agreement not checked here, only S-V is primary)
    return {'NUM': v_features['NUM']}

def vp_v_agreement_check(v_features):
    # VP inherits number from V
    return {'NUM': v_features['NUM']}

grammar = {
    'S': [
        (['NP', 'VP'], s_agreement_check)
    ],
    'NP': [
        (['Det', 'N'], np_det_n_agreement_check),
        (['N'], np_n_agreement_check)
    ],
    'VP': [
        (['V', 'NP'], vp_v_np_agreement_check), # V with object
        (['V'], vp_v_agreement_check)          # V without object
    ]
}

# Test sentences
sentences_for_agreement = [
    "the cat runs",         # Accepted: 'the' (any), 'cat' (singular), 'runs' (singular)
    "a cat runs",           # Accepted: 'a' (singular), 'cat' (singular), 'runs' (singular)
    "the cats run",         # Accepted: 'the' (any), 'cats' (plural), 'run' (plural)
    "a cats run",           # Rejected: 'a' (singular) with 'cats' (plural)
    "the cat run",          # Rejected: 'cat' (singular) with 'run' (plural)
    "the cats runs",        # Rejected: 'cats' (plural) with 'runs' (singular)
    "men eat",              # Accepted: 'men' (plural), 'eat' (plural)
    "man eats",             # Accepted: 'man' (singular), 'eats' (singular)
    "the dogs chase the cat", # Accepted
    "the dog chases the cats"  # Accepted
]

feature_parser = FeatureParser(grammar, lexicon)

print("Agreement Checking Results:")
for sentence in sentences_for_agreement:
    print(f"\nSentence: '{sentence}'")
    parse_tree = feature_parser.parse(sentence)
    if parse_tree:
        print("Accepted. Parse Tree with Features:")
        print(parse_tree)
    else:
        print("Rejected. Agreement or parsing failed.")
```

15. Implement probabilistic context-free grammar parsing for a sentence using python.

### Probabilistic Context-Free Grammar (PCFG) Parsing

A Probabilistic Context-Free Grammar (PCFG) is a context-free grammar where each production rule has a probability. These probabilities allow the parser to rank different possible parse trees for an ambiguous sentence and choose the most likely one. The CKY (Cocke-Kasami-Younger) algorithm can be adapted to find the most probable parse for a sentence given a PCFG.

```python
import collections

class PCFGParser:
    def __init__(self, pcfg_rules, start_symbol='S'):
        # pcfg_rules is a dictionary: non_terminal -> list of (probability, production_tuple)
        # Example: {'S': [(1.0, ('NP', 'VP'))], ...}
        self.pcfg_rules = pcfg_rules
        self.start_symbol = start_symbol

    def parse(self, sentence_str):
        words = sentence_str.split()
        n = len(words)

        # Initialize the CKY chart
        # chart[i][j] stores a dict: NonTerminal -> (max_prob, backpointer)
        # backpointer = ((split_point, left_NT, right_NT) or (unary_NT,) or word)
        chart = [[collections.defaultdict(lambda: (0.0, None)) for _ in range(n + 1)] for _ in range(n + 1)]

        # Fill in the diagonal (i.e., rules matching single words)
        for i in range(n):
            word = words[i]
            for lhs, productions in self.pcfg_rules.items():
                for prob, rhs in productions:
                    if len(rhs) == 1 and rhs[0] == word: # Terminal production A -> a
                        if prob > chart[i][i+1][lhs][0]:
                            chart[i][i+1][lhs] = (prob, word) # Backpointer is the word itself

        # Fill in the rest of the chart
        for length in range(2, n + 1):  # length of span
            for i in range(n - length + 1):  # start of span
                j = i + length  # end of span

                # Step 1: Process binary rules (A -> B C)
                for k in range(i + 1, j):  # split point
                    for lhs_A, prods_A in self.pcfg_rules.items():
                        for prob_A, rhs_A in prods_A:
                            if len(rhs_A) == 2:  # Binary rule A -> B C
                                B, C = rhs_A
                                prob_B = chart[i][k][B][0] # Probability of B spanning (i, k)
                                prob_C = chart[k][j][C][0] # Probability of C spanning (k, j)

                                if prob_B > 0 and prob_C > 0:
                                    current_prob = prob_A * prob_B * prob_C
                                    if current_prob > chart[i][j][lhs_A][0]:
                                        chart[i][j][lhs_A] = (current_prob, (k, B, C)) # Backpointer for binary rule

                # Step 2: Process unary rules (A -> B) - iterate until no new entries can be added
                changed = True
                while changed:
                    changed = False
                    # Iterate over existing non-terminals in chart[i][j]
                    # We need to iterate over a copy of the keys to allow modification during iteration
                    current_nts_in_cell = list(chart[i][j].keys())

                    for existing_nt in current_nts_in_cell:
                        existing_prob, _ = chart[i][j][existing_nt] # Get prob of existing_nt

                        # Look for unary rules A -> existing_nt
                        for lhs_A, prods_A in self.pcfg_rules.items():
                            for prob_A, rhs_A in prods_A:
                                if len(rhs_A) == 1 and rhs_A[0] == existing_nt: # Unary rule A -> B
                                    # Ensure we don't apply A -> A recursively if it exists
                                    if lhs_A == existing_nt:
                                        continue

                                    current_prob = prob_A * existing_prob
                                    if current_prob > chart[i][j][lhs_A][0]:
                                        chart[i][j][lhs_A] = (current_prob, (existing_nt,)) # Backpointer for unary rule (tuple of 1)
                                        changed = True

        # Reconstruct the most probable parse tree if a parse exists for the start symbol
        max_prob, backpointer = chart[0][n][self.start_symbol]
        if max_prob > 0:
            return max_prob, self._reconstruct_tree(self.start_symbol, 0, n, chart)
        return 0.0, None

    def _reconstruct_tree(self, lhs, i, j, chart):
        max_prob, backpointer = chart[i][j][lhs]

        if isinstance(backpointer, str):  # Terminal node (backpointer is the word)
            # The prob attribute for the ParseTreeNode of the non-terminal (lhs) should reflect
            # the probability of the rule application itself.
            # The actual word node will have a prob=1.0 as it's a leaf.
            return ParseTreeNode(lhs, [ParseTreeNode(backpointer, prob=1.0)], prob=max_prob)
        elif isinstance(backpointer, tuple) and len(backpointer) == 1: # Unary non-terminal rule (e.g., ('N',))
            child_nt = backpointer[0]
            child_node = self._reconstruct_tree(child_nt, i, j, chart) # Unary rule doesn't change span (i,j)
            return ParseTreeNode(lhs, [child_node], prob=max_prob)
        else:  # Binary non-terminal rule (k, B, C)
            k, B, C = backpointer
            left_child = self._reconstruct_tree(B, i, k, chart)
            right_child = self._reconstruct_tree(C, k, j, chart)
            return ParseTreeNode(lhs, [left_child, right_child], prob=max_prob)

class ParseTreeNode:
    def __init__(self, label, children=None, prob=1.0):
        self.label = label
        self.children = children if children is not None else []
        self.prob = prob

    def __str__(self, level=0):
        ret = "  " * level + f"{self.label} (p={self.prob:.4f})\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret

    def __repr__(self):
        return f"Node({self.label}, {self.prob}, {self.children})"

# --- Define a sample PCFG ---
# Grammar rules are (probability, production_tuple)
# Note: For simplicity, all word productions (e.g., N -> cat) are assumed to have a probability of 1.0
# when they match directly from the lexicon. A more robust PCFG would include these probabilities.

pcfg_grammar = {
    'S': [
        (1.0, ('NP', 'VP'))
    ],
    'NP': [
        (0.6, ('Det', 'N')),
        (0.4, ('N'))
    ],
    'VP': [
        (0.7, ('V', 'NP')),
        (0.3, ('V'))
    ],
    'Det': [
        (0.6, ('the')),
        (0.4, ('a'))
    ],
    'N': [
        (0.5, ('cat')),
        (0.3, ('dog')),
        (0.2, ('man'))
    ],
    'V': [
        (0.5, ('chased')),
        (0.5, ('ate'))
    ]
}

# Test sentences
sentences_for_pcfg = [
    "the cat chased the dog",
    "a man ate",
    "cat chased dog",
    "the dog"
]

pcfg_parser = PCFGParser(pcfg_grammar)

print("PCFG Parsing Results (Most Probable Parse Tree):\n")
for sentence in sentences_for_pcfg:
    print(f"Sentence: '{sentence}'")
    max_prob, parse_tree = pcfg_parser.parse(sentence)
    if parse_tree:
        print(f"  Accepted with probability: {max_prob:.6f}")
        print("  Parse Tree:")
        print(parse_tree)
    else:
        print("  Rejected. No parse found.")
```
